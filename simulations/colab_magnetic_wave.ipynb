```python
import numpy as np
import matplotlib.pyplot as plt
import plotly.graph_objects as go

# --- COLAB SIMULATION: 3-PHASE TRAVELING WAVE ---
# This visualizes the theoretical B-field generated by the 3 PCB phases 
# over space (x) and time (t).

# Spatial parameters (the length of our PCB)
x = np.linspace(0, 0.2, 500) # 20cm PCB
tau_p = 0.015 # Pole pitch
k = np.pi / tau_p

# Time parameter
f = 100 # Hz
omega = 2 * np.pi * f
t = 0.002 # Snapshot at 2 milliseconds

# 3-Phase PCB Trace Currents (120 degrees apart)
# Phase A: 0 rad, Phase B: 2pi/3 rad, Phase C: 4pi/3 rad
phase_A = np.cos(k*x) * np.cos(omega*t)
phase_B = np.cos(k*x - 2*np.pi/3) * np.cos(omega*t - 2*np.pi/3)
phase_C = np.cos(k*x - 4*np.pi/3) * np.cos(omega*t - 4*np.pi/3)

# By superposition, the total magnetic field summates into a traveling wave!
B_total = phase_A + phase_B + phase_C

# --- PLOTTING ---
plt.figure(figsize=(12, 5))
plt.plot(x, phase_A, 'r--', alpha=0.5, label='Phase A Trace Field')
plt.plot(x, phase_B, 'g--', alpha=0.5, label='Phase B Trace Field')
plt.plot(x, phase_C, 'b--', alpha=0.5, label='Phase C Trace Field')
plt.plot(x, B_total, 'w-', linewidth=3, label='Resultant Traveling Wave')

plt.title(f'PCB Magnetic Flux Density (Snapshot at t={t}s)', fontsize=14)
plt.xlabel('Distance along PCB (m)')
plt.ylabel('Normalized Magnetic Flux Density (B_z)')
plt.legend(loc='upper right')
plt.grid(True, alpha=0.2)
# plt.style.use('light_background') # Uncomment if running in Colab dark mode
plt.show()
```
